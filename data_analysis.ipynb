```python
import matplotlib.pyplot as plt
import numpy as np
import json
import os
import sys
import glob
```

```python
# define paths
plot_dir = f"plots/td3"
metric_dir = f"metrics/td3/*.json"
os.makedirs(plot_dir, exist_ok=True)
```

```python
# get data
metric_files = glob.glob(metric_dir)
hopper_dict = {}
ant_dict = {}
cheetah_dict = {}
for file in metric_files:
    with open(file, 'r') as f:
        data = json.load(f)

    if data['env_name'] == 'Hopper-v5': hopper_dict = data
    if data['env_name'] == 'HalfCheetah-v5': cheetah_dict = data
    if data['env_name'] == "Ant-v5": ant_dict = data
```

```python
# plot hopper 
plt.figure(figsize=(10,5))
plt.plot(hopper_dict["steps"], hopper_dict["rewards"], label="Hopper Curve")
plt.xlabel("Steps")
plt.ylabel("Rewards")
plt.title("TD3 on Hopper")
```

```python
# plot cheetah
plt.figure(figsize=(10,5))
plt.plot(cheetah_dict["steps"], cheetah_dict["rewards"], label="Cheetah Curve")
plt.xlabel("Steps")
plt.ylabel("Rewards")
plt.title("TD3 on HalfCheetah")
```

```python
# plot Ant 
plt.figure(figsize=(10,5))
plt.plot(ant_dict["steps"], ant_dict["rewards"], label="Ant Curve")
plt.xlabel("Steps")
plt.ylabel("Rewards")
plt.title("TD3 on Ant")
```
